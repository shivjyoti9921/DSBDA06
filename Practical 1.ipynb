```python
# -----------------------------------------------------------
# Non-Recursive Fibonacci Program using variables a and b
# -----------------------------------------------------------

# Take input from the user
n = int(input("Enter number of terms: "))

# Check if the entered number is valid
if n <= 0:
    print("Please enter a positive integer")
else:
    # Initialize first two Fibonacci numbers
    a, b = 0, 1

    print("Fibonacci sequence (Iterative):")
    
    # Loop to generate and print Fibonacci terms
    for i in range(n):
        # Print the current Fibonacci number
        print(a, end=" ")

        # Update 'a' and 'b' for the next iteration
        # 'a' becomes the previous 'b'
        # 'b' becomes the sum of the previous 'a' and 'b'
        a, b = b, a + b
```

```python
# -----------------------------------------------------------
# Recursive Fibonacci Program using variables a and b
# -----------------------------------------------------------

# Define a recursive function to print Fibonacci numbers
def fibonacci(a, b, n):
    # Base condition: if n > 0, continue printing
    if n > 0:
        # Print the current Fibonacci term
        print(a, end=" ")

        # Recursive call with:
        # 'a' replaced by 'b' (next term)
        # 'b' replaced by 'a + b' (sum of previous two terms)
        # and 'n' reduced by 1
        fibonacci(b, a + b, n - 1)

# Take input from the user
n = int(input("Enter number of terms: "))

# Check for valid input
if n <= 0:
    print("Please enter a positive integer")
else:
    # Print message and start recursion with first two Fibonacci numbers
    print("Fibonacci sequence (Recursive):")
    fibonacci(0, 1, n)
```

```python
# Non-recursive program to display the Fibonacci sequence up to n-th term

nterms = int(input("How many terms? "))

# first two terms
n1, n2 = 0, 1
count = 0

# check if the number of terms is valid
if nterms <= 0:
    print("Please enter a positive integer")
# if there is only one term, return n1
elif nterms == 1:
    print("Fibonacci sequence up to", nterms, ":")
    print(n1)
# generate fibonacci sequence
else:
    print("Fibonacci sequence:")
    while count < nterms:
        print(n1)
        nth = n1 + n2
        # update values
        n1 = n2
        n2 = nth
        count += 1
```

```python
# Recursive program to display the Fibonacci sequence

def recur_fibo(n):
    if n <= 1:
        return n
    else:
        return recur_fibo(n - 1) + recur_fibo(n - 2)


nterms = 7

# check if the number of terms is valid
if nterms <= 0:
    print("Please enter a positive integer")
else:
    print("Fibonacci sequence:")
    for i in range(nterms):
        print(recur_fibo(i))
```
